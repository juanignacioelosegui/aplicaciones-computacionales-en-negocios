### Ejercicio 1: Variable aleatoria discreta

1) La variable aleatoria discreta \( X \) toma valores en el conjunto {1, 2, 3, 4} con probabilidades:

$$
P[X = i] = i \cdot c
$$
$$
\quad \text{para } i = 1, 2, 3, 4 \text{ y alguna constante } c.
$$

a) ¿Cuál es el valor de $ c $?

b) Graficar la función la de probabilidad (gráfico de barras)


```python
import matplotlib.pyplot as plt

# Valores posibles de X
x_vals = [1, 2, 3, 4]

#P[X=i] = i*c, debe sumar 1 -> c*(1+2+3+4)=1 -> c=1/10
c = 1/10

# Probabilidades P[X = i] = i * c
probas = [i * c for i in x_vals]
print("c =", c)
print("Probabilidades:", probas)

#Grafico de barras
plt.bar(x_vals, probas, tick_label=x_vals)
plt.xlabel("Valores de X")
plt.ylabel("P[X = x]")
plt.title("Funci\u00f3n de probabilidad de X")
plt.grid(axis='y', linestyle='--', alpha=1)
plt.show()
```

c) Graficar la función acumulada

```python
import matplotlib.pyplot as plt
import numpy as np

# Valores posibles de X
x = np.array([1, 2, 3, 4])

c = 1/10

#Probabilidades
p = x * c

F = np.cumsum(p)  # ACUMULADA

print("p =", p)
print("F =", F)

plt.figure(figsize=(6,4))
plt.step(x, F, where='post')
plt.title('Funci\u00f3n de distribuci\u00f3n acumulada')
plt.xlabel('Valor de X')
plt.ylabel('F(X)')
plt.ylim(0, 1.05)
plt.grid(True)
plt.show()
```

d) ¿Cuánto vale la siguiente probabilidad?
$$
P(2 \leq X \leq 3)
$$

**Resolución:** $P(2\le X\le 3) = P(X=2)+P(X=3) = 2c+3c = 0.2+0.3 = \mathbf{0.5}$

$$
\mathbb{E}[X] = \sum_{i=1}^{4} i\cdot P[X=i] = 1(0.1)+2(0.2)+3(0.3)+4(0.4) = \mathbf{3.0}

$$
\text{Var}(X) = \mathbb{E}[X^2]-\mathbb{E}[X]^2 = (1^2\cdot0.1+2^2\cdot0.2+3^2\cdot0.3+4^2\cdot0.4) - 3.0^2 = 10.0-9.0=\mathbf{1.0}
$$

### Ejercicio 2: Variable aleatoria continua

La variable aleatoria continua $X$ tiene una función de densidad de probabilidad dada por

$$
f(x) = cx, \quad 0 < x < 1
$$


(a) Encontrar la constante $c$.

(b) Graficar la distribución de la probabilidad.

```python
import numpy as np
import matplotlib.pyplot as plt

c = 2

# Valores de x
x = np.linspace(0, 1, 200)

# funcion de densidad puntual
f_x = c * x

# ACUMULADA: F(x) =  c*t dt = c*x^2/2 = x^2 (con c=2)

plt.figure(figsize=(6, 4))
plt.plot(x, f_x, label=r'f(x) = cx', color='green')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Función de distribución de probabilidad')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

c) ¿Cuánto es $ P(X > 0.5)$?

**Solución:** $$P(X>0.5) = \int_{0.5}^1 f(x)\,dx = 2\int_{0.5}^1 x\,dx = 1-0.5^2 = 1-0.25 = \mathbf{0.75}$$

$ c)\quad Calcular \quad \mathbb{E}[X] $

**Resolución:**
$$
\mathbb{E}[X]=\int_0^1 x\,f(x)\,dx = \int_0^1 x\cdot 2x\,dx = 2\int_0^1 x^2\,dx = 2\cdot\frac{1}{3}=\mathbf{0.66}
$$

### Ejercicio 4

El universo de eventos posibles es un cuadrado y cualquier otro evento es
representado por un rectángulo incluido en el cuadrado. 

Si la probabilidad de cada evento es proporcional al área del rectángulo asociado, representar un ejemplo de
dos eventos A y B independientes entre sí.

---

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Universo: cuadrado unidad [0,1]x[0,1] con area total 1 = probabilidad total 1
# Evento A: franja VERTICAL en x, ancho 0.5  -> P(A) = 0.5
# Evento B: franja HORIZONTAL en y, alto 0.5 -> P(B) = 0.5
# Como A es una franja vertical (recorre todo el alto) y B una franja horizontal
# (recorre todo el ancho), su intersección es el rectángulo [0,0.5]x[0,0.5],
# cuya area es 0.5*0.5 = 0.25 = P(A)*P(B)  =>  A y B son INDEPENDIENTES

fig, ax = plt.subplots(figsize=(5,5))

# Cuadrado universo
ax.add_patch(patches.Rectangle((0,0), 1, 1, fill=False, edgecolor='black', lw=2))

# Evento A (franja vertical, azul)
ax.add_patch(patches.Rectangle((0,0), 0.5, 1, alpha=0.3, facecolor='blue', label='A (P=0.5)'))

# Evento B (franja horizontal, rojo)
ax.add_patch(patches.Rectangle((0,0), 1, 0.5, alpha=0.3, facecolor='red', label='B (P=0.5)'))

# Interseccion A y B (mas oscuro, se ve automaticamente por superposicion de colores)
ax.add_patch(patches.Rectangle((0,0), 0.5, 0.5, alpha=0.5, facecolor='purple', label='A\u2229B (P=0.25)'))

ax.set_xlim(0,1)
ax.set_ylim(0,1)
ax.set_aspect('equal')
ax.set_title('A y B independientes: P(A\u2229B) = P(A)\u00b7P(B) = 0.5\u00d70.5 = 0.25')
ax.legend(loc='upper right')
plt.show()

print("P(A) =", 0.5, " P(B) =", 0.5, " P(A)*P(B) =", 0.5*0.5, " P(A\u2229B) (area) =", 0.5*0.5)
```

### Ejercicio 5: Seguridad de aviones según cantidad de motores

Un avión necesita al menos la **mitad de sus motores funcionando** para poder volar.  
Cada motor funciona de manera independiente con probabilidad $ p $.

¿Para qué valores de $ p $ un avión con **tres motores** es más seguro que uno con **cinco motores**?

---

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Definimos las funciones de probabilidad
# Avion de 3 motores: necesita al menos 2 de 3 funcionando -> P(X>=2), X~Bin(3,p)
def P3(p):
    return 1 - binom.cdf(1, 3, p) 

# Avion de 5 motores: necesita al menos 3 de 5 funcionando -> P(X>=3), X~Bin(5,p)
def P5(p):
    return 1 - binom.cdf(2, 5, p)

#FUNCION PARA ENCONTRAR CRUCES 
def encontrar_cruces(x, y1, y2):
    # Diferencia entre las funciones
    diff = y1 - y2
    # Indices donde la diferencia cambia de signo
    indices_cruce = np.where(np.diff(np.sign(diff)))[0]
    
    return x[indices_cruce]

# Vector de valores p de 0 a 1
p = np.linspace(0, 1, 1000)

# Calculamos las probabilidades
prob3 = P3(p)
prob5 = P5(p)

# Buscamos los cruces
cruces = encontrar_cruces(p, prob3, prob5)

plt.figure(figsize=(8,5))
plt.plot(p, prob3, label='Avi\u00f3n 3 motores (>=2 funcionando)', color='blue', lw=2)
plt.plot(p, prob5, label='Avi\u00f3n 5 motores (>=3 funcionando)', color='red', lw=2)

#lineas verticales en los cruces encontrados
for cruce in cruces:
    plt.axvline(cruce, color='green', linestyle='--', label=f'Punto de cruce: {cruce:.1f}')

plt.title('Probabilidad de que el avi\u00f3n pueda volar seg\u00fan p')
plt.xlabel('Probabilidad p de que un motor funcione')
plt.ylabel('Probabilidad de poder volar')
plt.legend()
plt.grid(True)
plt.show()

print("Punto(s) de cruce encontrados:", cruces)
print()
print("Conclusión:")
print("- Para p < 0.5: P3(p) > P5(p)  -> el avión de 3 motores es MAS seguro")
print("- Para p > 0.5: P5(p) > P3(p)  -> el avión de 5 motores es MAS seguro")
print("- Para p = 0.5: ambos son igual de seguros (P3=P5=0.5)")
```

### Ejercicio 8: Retornos del índice Nasdaq

En la carpeta de **Trabajos Prácticos** del Campus Virtual hay un archivo con el índice **Nasdaq** durante los últimos 5 años.

Lectura de los datos

```python
import pandas as pd

datos = pd.read_csv("NASDAQ100.csv")
datos.head()
```

a) ¿Qué mide este índice?

El **retorno diario** de un activo financiero se define como la **diferencia porcentual** de precios entre dos días hábiles consecutivos:

$$
\text{Retorno}(t) = \frac{X(t) - X(t-1)}{X(t-1)}
$$

donde $ X(t) $ es el valor del activo en el día $ t $.

b) Generá la serie histórica de retornos del Nasdaq


```python
import pandas as pd

# Ordenar por fecha
datos = datos.sort_values('observation_date').reset_index(drop=True)

precios = datos['NASDAQ100'].values
retornos = [None]  # el primer dia no tiene retorno (no hay dia anterior)

# Retorno(t) = (X(t) - X(t-1)) / X(t-1)
for t in range(1, len(precios)):
    retornos.append((precios[t] - precios[t-1]) / precios[t-1])

datos['Retorno diario'] = retornos

print(datos)
```

c) Hacé un **histograma de los retornos diarios**. ¿A qué distribución se parece más?  
- Uniforme  
- Exponencial  
- Normal  


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, expon, uniform

retornos = datos['Retorno diario'].dropna()

# Histograma de los retornos
plt.hist(retornos, bins=50, density=True, alpha=0.6, color='g', label='Retornos diarios')

# Rango para ajustar las distribuciones
x = np.linspace(retornos.min(), retornos.max(), 1000)

# Ajustamos una Normal a los datos y la superponemos, para comparar visualmente
mu_fit, sigma_fit = norm.fit(retornos)
plt.plot(x, norm.pdf(x, mu_fit, sigma_fit), 'r-', lw=2, label=f'Normal ajustada (mu={mu_fit:.4f}, sigma={sigma_fit:.4f})')

plt.title('Histograma de retornos diarios y distribuciones ajustadas')
plt.xlabel('Retorno diario')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True)
plt.show()

# Respuesta: los retornos diarios de un indice bursatil suelen parecerse
# a una distribucion NORMAL (simetrica alrededor de 0, con colas mas pesadas
# de lo que predice la Normal pero de las 3 opciones,
# Normal es la que mejor aproxima).
```

d) ¿Cuál fue el **retorno diario medio** durante los últimos 5 años?

```python
import pandas as pd

# Por las dudas pasamos la fecha a una variable tipo datetime para buscar los 5 años
datos['observation_date'] = pd.to_datetime(datos['observation_date'])

retorno_medio = datos['Retorno diario'].mean()

print(f"Retorno diario medio ultimos 5 años: {retorno_medio:.6f}")
```

e) ¿Cuál fue la **desviación estándar** de los retornos diarios durante ese período?


```python

# Calculamos la desviacion estandar de los retornos diarios, excluyendo NaN
desviacion_estandar = datos['Retorno diario'].std()

print(f"Desviacion estandar de los retornos diarios ultimos 5 anios: {desviacion_estandar:.6f}")
```

f) ¿Cuál es la correlación entre retornos de días consecutivos?


```python
# Correlacion entre retornos de un dia y el dia siguiente
correlacion = datos['Retorno diario'].dropna().autocorr(lag=1)
print(f"Correlacion entre retornos de dias consecutivos: {correlacion:.6f}")
```
