### Caminata aleatoria en una cancha de fútbol

Comenzás a caminar desde el **centro del círculo central** de una cancha de fútbol profesional.  

- En cada segundo das:  
  - un paso de 1 metro en la dirección paralela al eje que une los dos arcos, y  
  - un paso de 1 metro en la dirección perpendicular.  
- Cada paso puede ser hacia adelante o hacia atrás con **igual probabilidad**, de manera independiente entre direcciones y pasos consecutivos.  

 Simula muchos caminos para responder con precisión:  
   - ¿Cuánto tardas en promedio en salir de la cancha?  
   - ¿En qué proporción de veces salís **entrando por uno de los dos arcos**?


```python
import numpy as np

largo = 105 # Eje x
ancho = 68  # Eje y
N = 300  # Aumentado para mejorar la precisión de la estimación
salidas_arco = 0

# S es un array para guardar los segundos que tarda cada una de las N caminatas
S = np.zeros(N)

for n in range(N):
    x = 0
    y = 0
    segundos = 0
    
    # Como hay un "break" más abajo que corta el ciclo, acá basta con poner True.
    # Alternativamente podrías poner: abs(x) < largo/2 and abs(y) < ancho/2
    while segundos <= 60**2 and True:
        segundos += 1
        
        # Paso aleatorio en el eje X e Y: usamos np.random.choice para elegir entre -1 y 1
        x = x + np.random.choice([-1, 1])
        y = y + np.random.choice([-1, 1])
        
        if abs(x) >= largo/2 or abs(y) >= ancho/2:
            # Guardamos el tiempo que tardó en salir
            S[n] = segundos

            # Chequeamos si salió por los bordes correspondientes al arco
            if abs(x) >= largo/2 and abs(y) <= 3.6:
                salidas_arco += 1

            break

tiempo_salida_promedio = S.sum() / N
proporcion = salidas_arco / N

print("En promedio tardan: ", tiempo_salida_promedio)
print("La proporción que entran al arco son: ", proporcion)

# Errores de estimación

# Error para la proporcion (variable Bernoulli)
# La desviación estándar de una distribución Bernoulli es la raíz de p*(1-p)
SE = np.sqrt(proporcion * (1 - proporcion))

# El error estándar de Monte Carlo divide la desviación estándar por la raíz de N
error_proporcion = SE / np.sqrt(N)

print("Error MC para la proporcion de salidas por arco: ", round(error_proporcion, 4))

# Error para el tiempo promedio
error_tiempo = np.std(S) / np.sqrt(N)
print("Error MC para el tiempo que tarda en salir: ", round(error_tiempo, 4))

# Interpretación
print("\nEs decir:")
print("La probabilidad de salir por el arco está aproximadamente entre ", proporcion, " ± ", round(error_proporcion, 4))
```
