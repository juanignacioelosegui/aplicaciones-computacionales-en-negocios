## Repaso

Para una variable aleatoria discreta $X$ con función de probabilidad $P[X=x_i]$ con $i=1,2,\dots,n$:
$$  \sum_{i=1}^{n}  P[X=x_i]  = 1$$

Para una variable aleatoria continua $X$ con función de densidad $f_X(x)$:
$$ \int_{\mathbb{R}}  f_X(x) dx = 1 $$

Por definición la función de distribución de una variable aleatoria $X$ evaluada en un número real $t$ es:
$$ F_X(t) = P[X \leq t] $$

Por lo tanto:

Para una variable aleatoria discreta $X$ con función de probabilidad $P[X=x]$ :
$$  F_X(t) = \sum_{x \leq t}  P[X=x]  $$

Para una variable aleatoria continua $X$ con función de densidad $f_X(x)$:
$$  F_X(t) = \int_{- \infty}^{t}  f_X(x) dx $$

### Algunas funciones

**La** función de densidad de la distribución gaussiana (o normal) con $\mathbb{E}[X]=\mu$ y varianza $ Var(X) = \sigma^2 $ se define como:

$$ f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \cdot e^{-\frac{(x - \mu)^2}{2\sigma^2}} $$


La función de densidad de la distribución uniforme en el intervalo $[a, b]$ se define como:

$$ f(x) = \begin{cases} \frac{1}{b-a}, & \text{si } a \leq x \leq b \\ 0, & \text{en otro caso} \end{cases} $$


La función de probabilidad de la distribución de Poisson con tasa $\lambda$ se define como:

$$ P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!} $$


### Esperanza

Para una variable aleatoria discreta $X$ con función de probabilidad $P[X=x_i]$ con $i=1,2,\dots,n$ , la esperanza se define como:
$$ E[X] = \sum_{i=1}^{n} x_i P[X=x_i] $$

Para una variable aleatoria continua $X$ con función de densidad $f_X(x)$ , el valor esperado se define como la integral de Riemann:
$$ E[X] = \int_{\mathbb{R}} x f_X(x) dx $$


### Propiedades

1. Propiedad del valor esperado de una constante: Si $c$ es una constante, entonces:
$$ E[c] = c $$

2. Linealidad: Para cualquier variable aleatoria $X$ y $Y$ y constantes $a$ y $b$, se cumple:
$$ E[aX + bY] = aE[X] + bE[Y] $$


3. Propiedad del valor esperado de una función: Para una función \(g(x)\), se cumple:
$$ E[g(X)] = \sum_{x} g(x) P[X=x] \quad \text{(caso discreto)} $$
$$ E[g(X)] = \int_{-\infty}^{\infty} g(x) f_X(x) dx \quad \text{(caso continuo)} $$


4. Valor esperado de una variable aleatoria no negativa: Si \(X\) es una variable aleatoria no negativa $(X \geq 0)$, entonces $E[X] \geq 0$


### Varianza

La varianza se define como $$\text{Var}(X) = E[(X - E[X])^2] = E[X^2] -E[X]^2 $$

Por lo tanto, para una variable aleatoria discreta $X$ con función de probabilidad $P[X=x_i]$ con $i=1,2,\dots,n$, la varianza es:
$$ \text{Var}(X) = \sum_{i=1}^{n} (x_i - E[X])^2 P[X=x_i] $$

Por lo tanto, para una variable aleatoria continua $X$ con función de densidad $f_X(x)$, la varianza se define como:
$$ \text{Var}(X) = \int_{\mathbb{R}} (x - E[X])^2 f_X(x) dx $$

### Repaso: Funciones Bernoulli y Binomial

#### Función Bernoulli

Una **variable aleatoria Bernoulli** modela un experimento con solo dos resultados posibles:

- Éxito (1) con probabilidad $p$  
- Fracaso (0) con probabilidad $1 - p$

#### Definición matemática

$ 
X = \begin{cases}
1, & \text{con probabilidad } p \\
0, & \text{con probabilidad } 1-p
\end{cases}
$

#### Función de masa de probabilidad

$
P(X = x) = p^x (1-p)^{1-x} \quad \text{para } x \in \{0,1\}
$

#### Propiedades

- Esperanza: $\mathbb{E}[X] = p$  
- Varianza: $\mathrm{Var}(X) = p(1-p)$

#### ¿Para qué sirve?

Modelar eventos que solo pueden pasar o no pasar, como:

- Tirar una moneda y que salga cara  
- Que un motor funcione o falle  
- Que un usuario haga clic en un anuncio


### Distribución Binomial

La **distribución Binomial** cuenta el número de éxitos en $n$ experimentos Bernoulli independientes.

#### Parámetros

- $n$: cantidad de ensayos  
- $p$: probabilidad de éxito en cada ensayo

#### Función de masa de probabilidad

$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}
$

donde

$
\binom{n}{k} = \frac{n!}{k!(n-k)!}
$

#### Propiedades

- Esperanza: $\mathbb{E}[X] = np$  
- Varianza: $\mathrm{Var}(X) = np(1-p)$

#### ¿Para qué sirve?

Contar cuántos éxitos hay en un conjunto de experimentos idénticos y independientes, por ejemplo:

- Número de caras al lanzar una moneda 10 veces  
- Número de motores que funcionan correctamente en un avión con varios motores  
- Número de clics en 100 anuncios mostrados a usuarios


### Teorema de Markov

Sea $X$ una variable aleatoria no negativa y sea $a > 0$.  
Entonces:

$$P(X \ge a) \le \frac{\mathbb{E}[X]}{a}$$



### Teorema de Chebyshev

Sea $X$ una variable aleatoria con media $\mu$ y varianza $\sigma^2$ finitas.  
Entonces, para cualquier $k > 0$:

$$
P(|X - \mu| \geq k\sigma) \leq \frac{1}{k^2}
$$

De forma equivalente:

$$
P(|X - \mu| < k\sigma) \geq 1 - \frac{1}{k^2}
$$
