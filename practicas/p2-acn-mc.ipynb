### Estimacion de Pi:

Vamos a estimar Pi con el método de MC

En este cálculo se tienen que hacer dos consideraciones importantes:

1. Si los puntos no están uniformemente distribuidos, el método es inválido.
2. La aproximación será pobre si solo se lanzan unos pocos puntos. En promedio, la aproximación mejora conforme se aumenta el número de puntos.

```python
import random

dentro_circulo = 0
N_puntos = 100

for n in range(N_puntos):
    # Generar punto aleatorio en [-1,1] x [-1,1]
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)

    if (x**2 + y**2) <= 1:
        dentro_circulo += 1

pi_estimado = 4 * dentro_circulo / N_puntos

print(pi_estimado)
```

```python
import numpy as np

def volumen_grilla(N, h=0.1):
    """
    Aproxima el volumen de la hiperesfera de radio = 1 en N dimensiones
    mediante una suma de Riemann N-dimensional.

    Idea: cubrimos el hipercubo [-1, 1]^N con una grilla dando pasos de a h (1/10).
    Cada punto de la grilla representa el centro de una celda de volumen (1/10)^N.
    Si sumamos el volumen de todas las celdas que caen dentro de la esfera, 
    obtenemos una aproximación al volumen real.

    Cuantos más puntos por eje (h más chico), más fina la grilla y mejor
    la aproximación, pero el costo crece exponencialmente.
    """

    #Discretizamos cada eje en el intervalo [-1, 1]
    valores_por_eje = np.arange(-1, 1 + h, h)

    m = len(valores_por_eje)  # cantidad de puntos por eje

    #Construimos la grilla como un array de N coordenadas
    coordenadas = np.meshgrid(*([valores_por_eje] * N), indexing="ij")

    #Calculamos ||x||^2 = x1^2 + x2^2 + ... + xN^2,
    #sumando coordenada a coordenada
    norma_cuadrada = np.zeros_like(coordenadas[0])
    for eje in coordenadas:
        norma_cuadrada += eje**2

    #Cuento si cae dentro de la esfera: ||x|| <= 1
    dentro_de_la_esfera = norma_cuadrada <= 1

    #Cada celda de la grilla tiene volumen h^N
    volumen_celda = h**N
    cantidad_de_puntos_dentro = np.sum(dentro_de_la_esfera)

    return volumen_celda * cantidad_de_puntos_dentro

# Nota: una grilla determinista necesita m^N puntos (m = cantidad de puntos por eje).
# Esto crece exponencialmente con N: para N=10 y h=0.1 serían 21^10 puntos, imposible de almacenar en memoria.

# Monte Carlo, en cambio, evalúa una cantidad de puntos elegida por vos independientemente de N

for N in [3, 5]:
    print(f"N = {N}")

    teorico = volumen_analitico(N)
    estimado = volumen_grilla(N)

    error_abs = abs(estimado - teorico)
    error_rel = error_abs / teorico

    print(f"  Analítico:        {teorico:.6f}")
    print(f"  Grilla (Riemann): {estimado:.6f}")
    print(f"  Error absoluto:   {error_abs:.6f}")
    print(f"  Error relativo:   {100*error_rel:.4f}%")
    print()
```
